## 3. Success metric

*One metric you can defend. What number means 'good'?*
My primary metric is ROC AUC on held-out clients (client-grouped split),
compared against a baseline of grouped summaries by position tier.
"Good" means AUC clearly above 0.5 and above the baseline, with the
result stable across folds. I do not treat a high correlation as success:
in noisy search data, honest effects are modest, and an unusually high
score would trigger a leakage check. Supporting evidence: effect sizes
and permutation importance per signal, and whether each signal's
direction is consistent across clients.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*
Hand-written if-statements get messy here because there are many signals,
the relationships are gradual rather than yes/no, and they interact with
each other. Rules also depend on hard-coded thresholds that must be chosen
by hand, treat near-identical pages differently at the cutoff, and may not
generalize across clients. A model learns these patterns and weights from
the data, and a simple rule-based baseline is still kept as the benchmark.
